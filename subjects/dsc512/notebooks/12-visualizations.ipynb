Making informative visualizations (also called plots) is one of the important tasks in data analysis and storytelling.

Creating visualizations may be a part of data exploration – identifying outliers, necessary data transformations, or insights on model building. Or may be the end-goal – data storytelling, creating dashboards, and decision support tools.
In this notebook, we will show some of the essential features of pandas that is involved with data visualizations. We will cover how to generate line plots, bar plots, histogram plots, scatter plots, and box plots.

Since most of the plotting functionality of `pandas` is wrapped around `matplotlib` we will also show in this notebook how to incorporate certain elements on our visualization using features of `matplotlib`. Additionally, we will also use some essential plotting features of the `seaborn` package which are quite routinary when it comes to data analysis and data science.


````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib import rcParams


rcParams.update({'figure.figsize': (10, 6), 'axes.spines.top': False,
                 'axes.spines.right': False, 'axes.labelsize': 14,
                 'axes.titlesize': 16, 'axes.titleweight': 'bold',
                 'figure.titlesize': 18, 'figure.titleweight': 'bold',
                 'lines.linewidth': 3})
````

## 0 Data Preparation

The data that we'll be using is the population data from the countries of the world collected by world bank from the census reports and other staistical publications from national offices of the respective countries.

We start our demonstration by a routinary data loading and cleaning procedures.

````python
data = pd.read_csv('data/pop.csv', skiprows=4, index_col=['Country Code'])
country_codes = data['Country Name']
pop_data = data.loc[:, data.columns.str.isnumeric()].T
pop_data.index = pop_data.index.astype(int)
````

## 1 Line Plots

````python
ph_pop = pop_data.loc[:, 'PHL']
ph_pop
````

Both the `Series` and `DataFrame` objects of `pandas` have the `plot` attribute for making some basic plots. By default, `plot()` makes a line plot.

````python
ph_pop.plot()
````

Notice that the `plot()` method returns a matplotlib axes which we can use to further customize our visualization. We can do things such as adding axes labels and a figure title to our plot.

````python
ax = ph_pop.plot()
fig = plt.gcf()

ax.set_xlabel("Year")
ax.set_ylabel("Popoulation")
fig.suptitle("Philippine Population from 1960 to 2022");
````

We can also change how the tick labels appear by defining a function formatter.

````python
from matplotlib.ticker import FuncFormatter

def millions(x, pos):
    return f"{(x * 1e-6):1.0f} M"
````

````python
ax = ph_pop.plot()
fig = plt.gcf()

ax.set_xlabel("Year")
ax.set_ylabel("Popoulation")
ax.yaxis.set_major_formatter(FuncFormatter(millions))
````

See the table below for a partial list of all the options available when using the `plot` method. We'll be showing how each option can be used but the rest is up to you to explore.

<img src="images/series-plot-methods.png" style="width: 65%;">

For example, we can change the line style of our line plot by specifying the style string to be passed to `matplotlib`.

````python
ax = ph_pop.plot(style='o-')
fig = plt.gcf()

ax.set_xlabel("Year")
ax.set_ylabel("Popoulation")
ax.yaxis.set_major_formatter(FuncFormatter(millions))
fig.suptitle("Philippine Population from 1960 to 2022");
````

Since data frames are two dimensional data, we have the option to specify which columns (and rows) to plot by first performing data selection. In our case, let's select all the data from the South East Asian countries.

````python
sea_countries = ['Brunei Darussalam', 'Myanmar', 'Cambodia', 'Timor-Leste',
                 'Indonesia', 'Lao PDR', 'Malaysia', 'Philippines',
                 'Singapore', 'Thailand', 'Viet Nam']
sea_pop = pop_data.loc[:, country_codes.isin(sea_countries)]
sea_pop
````

The `DataFrame`'s `plot` method plots each of its columns as a different line on the same subplot with a legend created automatically.

````python
sea_pop.plot()
````

And just like in `Series`, the generated plot usually returns a matplotlib axes object, which we can further tweak to customize our plot. Here, we add axes labels and move the legend to a suitable location.

````python
ax = sea_pop.plot()
fig = plt.gcf()

ax.set_xlabel("Year")
ax.set_ylabel("Popoulation")
ax.legend(loc='upper left', bbox_to_anchor=(1., 1.))
ax.yaxis.set_major_formatter(FuncFormatter(millions))
fig.suptitle("SEA Countries Population from 1960 to 2022");
````

`DataFrame` has a number of specific options which allow for some flexibility on how the columns are handled. For example, whether to plot them all in the same subplot or to create separate subplot. Here are some of the available options:

<img src="images/dataframe-plot-methods.png" style="width: 55%;">

Setting `subplots` to `True` will create several `matplotlib.Axes` which you can modify succeedingly.

````python
axes = sea_pop.plot(subplots=True)
fig = plt.gcf()
fig.set_size_inches(10, 12)

for ax in axes:
    ax.set_xlabel("Year")
    ax.set_ylabel("Pop")
    ax.legend(loc='upper left', bbox_to_anchor=(1., 1.), frameon=False)
    ax.yaxis.set_major_formatter(FuncFormatter(millions))
fig.suptitle("SEA Countries Population from 1960 to 2022");
````

Depending on how you want your visualization to look like, you may opt to add or hide some elements of the plot. Here, we forego adding y-labels to each subplot and instead just add a generic label for all as a figure text.

````python
axes = sea_pop.plot(subplots=True)
fig = plt.gcf()
fig.set_size_inches(10, 12)

for ax in axes:
    ax.set_xlabel("Year")
    ax.legend(loc='upper left', bbox_to_anchor=(1., 1.), frameon=False)
    ax.yaxis.set_major_formatter(FuncFormatter(millions))
plt.text(0.025, 0.5, "Population", fontsize=14, transform=fig.transFigure,
         rotation=90)
fig.suptitle("SEA Countries Population from 1960 to 2022", y=0.92);
````

You can opt to pass a specific `layout` for more controllability on how the subplots are made and arranged.

````python
axes = sea_pop.plot(subplots=True, layout=(6, 2))
fig = plt.gcf()
fig.set_size_inches(10, 12)

for ax in axes.flatten():
    ax.set_xlabel("Year")
    ax.yaxis.set_major_formatter(FuncFormatter(millions))

fig.suptitle("SEA Countries Population from 1960 to 2022");
````

Since the legends are practically useless here, we can forego plotting them and instead add an axes title for each subplot corresponding to the country code of the corresponding line plot.

````python
axes = sea_pop.plot(subplots=True, layout=(6, 2), legend=False)
fig = plt.gcf()
fig.set_size_inches(10, 12)

for i, ax in enumerate(axes.flatten()[:-1]):
    ax.set_xlabel("Year")
    ax.yaxis.set_major_formatter(FuncFormatter(millions))
    ax.set_title(sea_pop.columns[i], fontsize=14, fontweight='normal')
fig.suptitle("SEA Countries Population from 1960 to 2022");
````

We can further refine how the subplots are placed by specifying the specifications of the grid to be used in plotting.

````python
gridspec = {'hspace': 0.40, 'wspace': 0.20}
axes = sea_pop.plot(subplots=True, layout=(6, 2), legend=False)
fig = plt.gcf()
fig.set_size_inches(10, 12)

for i, ax in enumerate(axes.flatten()[:-1]):
    ax.set_xlabel("Year", fontsize=10)
    ax.yaxis.set_major_formatter(FuncFormatter(millions))
    ax.set_title(sea_pop.columns[i], fontsize=12, fontweight='normal')
fig.suptitle("SEA Countries Population from 1960 to 2022")
plt.subplots_adjust(**gridspec);
````

## 2 Bar Plots

Generating a bar plot can be achieved by setting the `kind` as `bar` on the `plot` method. In this case, the `Series` or the `DataFrame` index will be used as the `x` (for `bar`) or `y` (`barh`) ticks.

````python
ax = sea_pop.loc[2022].plot(kind='bar')
fig = plt.gcf()

ax.set_ylabel("Population")
ax.yaxis.set_major_formatter(FuncFormatter(millions))
fig.suptitle("SEA Countries Population for the Year 2022")
````

Ideally, you want the tick labels to be readable.

````python
ax = sea_pop.loc[2022].plot(kind='bar')
fig = plt.gcf()

plt.xticks(rotation=0)
ax.set_ylabel("Population")
ax.yaxis.set_major_formatter(FuncFormatter(millions))
fig.suptitle("SEA Countries Population for the Year 2022");
````

This is an example of plotting a horizontal bar plot. The `y` axis correspond to the `Series`' index.

````python
ax = sea_pop.loc[2022].plot(kind='barh')
fig = plt.gcf()

ax.set_xlabel("Population")
ax.xaxis.set_major_formatter(FuncFormatter(millions))
fig.suptitle("SEA Countries Population for the Year 2022");
````

You can also sort the `Series` beforehand before generating the plot. You can also set the `color` argument to change the color of each bar being plotted.

````python
ax = sea_pop.loc[2022].sort_values().plot(
    kind='barh', color=sns.color_palette('tab10'))
fig = plt.gcf()

ax.set_xlabel("Population")
ax.xaxis.set_major_formatter(FuncFormatter(millions))
fig.suptitle("SEA Countries Population for the Year 2022");
````

Using the native `bar` or `barh` of matplotlib has its advantage if you want to easily annotate the numerical value of each bar.

````python
fig, ax = plt.subplots()

rects = ax.barh(sea_pop.loc[2022].sort_values().index,
                sea_pop.loc[2022].sort_values(),
                color=sns.color_palette('tab10'))
ax.bar_label(rects, padding=3, fmt=lambda x: f"{int(x):,}")
ax.set_xlabel("Population")
ax.xaxis.set_major_formatter(FuncFormatter(millions))
fig.suptitle("SEA Countries Population for the Year 2022");
````

An improvement of our current plot might be to change the location of the ticklabels, which we can do by setting the `tick_params` of the axes.

````python
fig, ax = plt.subplots()

rects = ax.barh(sea_pop.loc[2022].sort_values().index,
                sea_pop.loc[2022].sort_values(),
                color=sns.color_palette('tab10'))
ax.bar_label(rects, padding=3, fmt=lambda x: f"{int(x):,}")
ax.xaxis.set_major_formatter(FuncFormatter(millions))
ax.spines['top'].set_visible(True)
ax.spines['bottom'].set_visible(False)
ax.tick_params(labeltop=True, top=True, labelbottom=False, bottom=False)
fig.suptitle("SEA Population for the Year 2022", y=1.025);
````

With a `DataFrame`, bar plots group the values in each row in bars, side by side, for each value.

````python
sea_pop.iloc[-5:].T
````

````python
ax = sea_pop.iloc[-5:].T.plot(kind='bar')
fig = plt.gcf()

fig.suptitle("SEA Population for 2018 to 2022")
ax.set_ylabel("Population")
ax.legend(loc='upper left', bbox_to_anchor=(1., 1.))
ax.yaxis.set_major_formatter(FuncFormatter(millions))
plt.xticks(rotation=0);
````

We can also stack this instead, if we want to emphasize that there is a cumulative relationship between each bar on the same group.

````python
sea_change = sea_pop.iloc[-6:, [1, 5, 6, 8]].diff().dropna().T
sea_change
````

````python
ax = sea_change.plot(
    kind='barh', stacked=True, color=sns.color_palette('viridis'))
fig = plt.gcf()

ax.legend(loc='upper left', bbox_to_anchor=(1., 1.))
ax.set_xlabel("Population")
fig.suptitle("Change in Population Stacked from 2018 to 2022");
````

Usually, this kind of visualization is also useful, if the total value of each bar sums to 1. Allowing us to visualize the fraction of each bar.

````python
sea_change.T / sea_change.sum(axis='columns')
````

````python
sea_prop_change = (sea_change.T / sea_change.sum(axis='columns')).T
````

````python
ax = sea_prop_change.plot(
    kind='bar', stacked=True, color=sns.color_palette('plasma'))
fig = plt.gcf()

ax.legend(loc='upper left', bbox_to_anchor=(1., 1.))
plt.xticks(rotation=0)
fig.suptitle("Fractional Change in Population during 2018 to 2022");
````

## 3 Histogram and Density Plots

A *histogram* is a specific kind of a bar plot that gives a discretized display of value of frequency. The data points are split into discrete, evenly spaced bins, adn the number of data points in each bin is plotted.

For example, if we'd like to visualize the distribution of the relative growth rate for the year 2022 for all countries, we first compute the relative growth rate from the data.

````python
rgr = pop_data.diff() / pop_data
````

Then plot the histogram accordingly. We can adjust how the histogram is computed using the `bins` parameter, and we can adjust the spacing of each bar using the `rwidth` parameter.

````python
ax = rgr.loc[2022].plot(kind='hist', bins=50, rwidth=0.7)
fig = plt.gcf()

ax.set_xlabel("Relative Growth Rate")
fig.suptitle("Relative Growth Rate of World Population for the Year 2022");
````

Usually, we want to annotate points of interest in our plot. For example, the mean value of the distribution. We can do this by plotting a line along the mean, then adding an anotation to it.

````python
rgr_mean = rgr.loc[2022].mean()

ax = rgr.loc[2022].plot(kind='hist', bins=50, rwidth=0.7)
fig = plt.gcf()

ax.axvline(rgr_mean, color='tab:orange', ls='--')
ax.set_xlabel("Relative Growth Rate")
fig.suptitle("Relative Growth Rate of World Population for the Year 2022");
````

To properly place the annotating text, we need to use a blended transform so that the x coordinate is based on the data, while the y coordinate of the text follows the axes coordinates.

````python
from matplotlib.transforms import blended_transform_factory
````

````python
rgr_mean = rgr.loc[2022].mean()

ax = rgr.loc[2022].plot(kind='hist', bins=50, rwidth=0.7)
fig = plt.gcf()

ax.axvline(rgr_mean, color='tab:orange', ls='--')
ax.set_xlabel("Relative Growth Rate")
ax.text(
    rgr_mean + 0.003, 0.95, f"mean: {rgr_mean:.5f}",
    transform=blended_transform_factory(ax.transData, ax.transAxes),
    weight='bold', color='tab:orange'
)
fig.suptitle("Relative Growth Rate of World Population for the Year 2022");
````

If we prefer to plot a density plot using [kernel density estimation](https://en.wikipedia.org/wiki/Kernel_density_estimation), we can set the `kind` as `kde` instead.

````python
rgr_mean = rgr.loc[2022].mean()

ax = rgr.loc[2022].plot(kind='kde')
fig = plt.gcf()

ax.axvline(rgr_mean, color='tab:orange', ls='--')
ax.set_xlabel("Relative Growth Rate")
ax.text(
    rgr_mean + 0.003, 0.975, f"mean: {rgr_mean:.5f}",
    transform=blended_transform_factory(ax.transData, ax.transAxes),
    weight='bold', color='tab:orange'
)
fig.suptitle("Relative Growth Rate of World Population for the Year 2022");
````

## 4 Scatter or Point Plots

Point plots or scatter plots is a useful method to show the relationship between two one-dimensional data series. As a demonstration of creating scatter plots, we use this [data](https://www.kaggle.com/datasets/joebeachcapital/world-population-by-country-2023/) which are economic features based on the latest United Nations Population Division estimate.

Our dataset contains the following features:

```
Country - Name of countries and dependent territories.
Population2023 - Population in the year 2023
YearlyChange - Percentage Yearly Change in Population
NetChange - Net Change in Population
Density(P/Km²)- Population density (population per square km)
Land Area(Km²) - Land area of countries / dependent territories.
Migrants(net) - Total number of migrants
Fert.Rate - Fertility rate
Med.Age - Median age of the population
UrbanPop%- Percentage of urban population
WorldShare - Population share
```

We perform a quick data cleaning,

````python
pop_2023 = pd.read_csv(
    'data/pop2023.csv', na_values='N.A.', index_col='Country')
pop_2023
````

And we change the percentages value to float format.

````python
pop_2023.loc[:, 'YearlyChange'] = pop_2023.loc[:, 'YearlyChange'].str.replace(' %', '')
pop_2023.loc[:, 'UrbanPop%'] = pop_2023.loc[:, 'UrbanPop%'].str.replace(' %', '')
pop_2023.loc[:, 'WorldShare'] = pop_2023.loc[:, 'WorldShare'].str.replace(' %', '')
pop_2023.isetitem(
    np.arange(len(pop_2023.columns))[pop_2023.dtypes == 'object'],
    pop_2023.loc[:, pop_2023.dtypes == 'object'].astype(float).map(lambda x: x/100))
pop_2023
````

When using the `plot` method of the `DataFrame`, we specify which we plot as our x-coordinate and y-coordinate using the keywords `x` and `y`.

````python
ax = pop_2023.plot(kind='scatter', x='Density(P/Km²)', y='UrbanPop%')
ax.set_xscale('log')
````

Then, by utilizing the `ax` parameter of the `plot()` method, we can create subplots consisting of combinations of different features in our data frame.

````python
fig, axes = plt.subplots(2, 2, figsize=(12, 12))

pop_2023.plot(kind='scatter', x='Population2023', y='UrbanPop%', ax=axes[0][0])
pop_2023.plot(kind='scatter', x='Migrants(net)', y='UrbanPop%', ax=axes[0][1])
pop_2023.plot(kind='scatter', x='Fert.Rate', y='NetChange', ax=axes[1][0])
pop_2023.plot(kind='scatter', x='Density(P/Km²)', y='UrbanPop%', ax=axes[1][1])

axes[0][0].set_xscale('log')
axes[0][1].set_xscale('log')
axes[1][1].set_xscale('log')
plt.subplots_adjust(hspace=0.30)
````

A quick and easy way to plot each and every combination of pairs from our features is to use seaborn's `pair_plot`. Let's consider limiting the features to the first 4 features in the dataset.

````python
pop_2023.iloc[:, 1:5]
````

Seaborn places the histogram or density estimates of each variable along the diagonal while all other facets of the plot contain a scatter plot of the pair of the features.

````python
axes = sns.pairplot(pop_2023.iloc[:, 1:5])
axes.axes[0][0].set_yscale('log')
axes.axes[-1][0].set_xscale('log')
axes.axes[-1][0].set_yscale('log')
axes.axes[-1][-1].set_xscale('log')
````

## 5 Box Plots

When we want to compare the difference in distribution of a specific feature on different groups, a boxplot is a good visualization choice. As an example, let's look at the GDP per capita data for countries of the world for several years.

````python
gdp = pd.read_csv('data/gdp_cleaned.csv', index_col='Year')
gdp
````

Choosing the SEA countries, we can compare the distribution of GDP per capita (per year) of each of the SEA countries.

````python
fig, ax = plt.subplots(figsize=(12, 5))

gdp.loc[:, gdp.columns.isin(sea_countries)].plot(kind='box', ax=ax)
plt.xticks(rotation=45, ha='right')
ax.set_ylabel("GDP Per Capita (in USD)")
ax.set_yscale('log')
````

An alternative way of doing this is thru the use of seaborn's boxplot function.

````python
ax = sns.boxplot(data=gdp.loc[:, gdp.columns.isin(sea_countries)])
fig = plt.gcf()
fig.set_size_inches(12, 5)
plt.xticks(rotation=45, ha='right')
ax.set_ylabel("GDP Per Capita (in USD)")
ax.set_yscale('log')
````
