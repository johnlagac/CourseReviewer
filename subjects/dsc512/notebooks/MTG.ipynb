```python
"""
Demo of Python classes with Magic the Gathering (MTG) cards. 

All MTG cards have a name, mana cost, card type, (effect) text, and flavor text.
There are others I have left off (artist, print year, etc.) so you can practice on those. 
So the first class is the base class. 
After that, cards are divided into creatures and spells. 

The creature class inherits the base class (i.e. it also has name, mana cost, etc.) via the 'super' function
This class also gains properties specific to creatures (power and toughness values). 

The spell class similarly inherits the base class values, and gains a property specific to spells (instant or sorcery speed).
"""

class mtg_card:
    def __init__(self, name, mana_cost, card_type, text, flavor):
        self.name = name
        self.mana_cost = mana_cost
        self.card_type = card_type
        self.text = text
        self.flavor = flavor

    def __str__(self):
        return f"{self.name} [{self.mana_cost}]\n{self.card_type}\n{self.text}\n{self.flavor}"

# 
class mtg_creature(mtg_card):
    def __init__(self, name, mana_cost, card_type, text, flavor, power, toughness):
        super().__init__(name, mana_cost, card_type, text, flavor)
        self.power = power
        self.toughness = toughness

    def __str__(self):
        base_str = super().__str__()
        return f"{base_str}\n{self.power}/{self.toughness}"


class mtg_spell(mtg_card):
    def __init__(self, name, mana_cost, card_type, text, flavor, spell_type):
        super().__init__(name, mana_cost, card_type, text, flavor)
        self.spell_type = spell_type # either instant or sorcery

    def __str__(self):
        base_str = super().__str__()
        return f"{base_str}\nType: {self.spell_type}"
```

```python
# Example translation for Baneslayer Angel
creature = mtg_creature(name="Baneslayer Angel",
                        mana_cost="3WW",
                        card_type="Creature — Angel",
                        text="Flying, first strike, lifelink, protection from Demons and from Dragons",
                        flavor = "",
                        power=5,
                        toughness=5,
                        )

print(creature)
```
