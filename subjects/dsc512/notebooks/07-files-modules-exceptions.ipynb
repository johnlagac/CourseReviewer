## 1 Files

Python has built-in file object that allow you to read and write text files. We can initialize such objects using the built-in `open` function.

In order, to open a file, you need to tell where that program is. By default, Python assumes that the file you want to read is in the same directory as the program that is doing the reading. As such, if our file is in another folder, we need to refer to it via absolute pathing or relative pathing.

```python
file = open('data/file_example.txt', 'r')
contents = file.read()
file.close()
print(contents)
```

The second parameter in the `open` function specifies the mode in which you are opening the file. It can be:

| Mode | Value |
| ---  | ----- |
| Read | `r` | 
| Writing | `w` | 
| Exclusive Creation | `x` |
| Appending | `a` |
| Updating | `+` |
| Bytes | `b` |

We will demonstrate the use-case of each of the following mode in this notebook. But first, we discuss an important pattern when working with files, that is, using a context manager `with` statement.

### Using a `with` statement

A common programming pattern is to get access to a resource, do something with that resource, and then tidy up and release the resource. In our previous example, we read the file contents, then we tidied up by closing the file. However, if in between the reading and closing process, an error occurs, the `file.close()` will not be executed and the associated resources are never released.

Through the use of the `with` statement, we perform the `file.close()` process regardless of whether an error occurs. For this reason, the `with` statement is frequently used for file access.

```python
with open('data/file_example.txt', 'r') as file:
    contents = file.read()
print(contents)
```

Under the hood, when we use the `with` statement, the objects's `__enter__` method is called throuh which the result of the call is assigned to the variable (in this case as `file`). After the block has been executed, the `__exit__` method of the object is called. In this case, the `file` object closes the file.

### Techniques in reading files

When reading files, note that we always start at the current file cursor. This allows us to combine techniques that we want to use when reading files.

#### Reading the whole content or a specific number of characters.

When called with no arguments, `read` everything from the current file cursor all the way to the end of the file and moves the file cursor to the end of the file. With an argument, it reads that many characters and moves the file cursor after the characters that were just read.

```python
with open('file_example.txt', 'r') as file:
    first_ten_chars = file.read(10)
    the_rest = file.read()

print("The first 10 characters: ", first_ten_chars)
print("The rest of the file: ", the_rest)
```

#### Making a `list` of lines from a file

If you want to et a Python list of strings containing the individual lines from a file, the function `readlines` work like `read`, but it splits up the lines into a list of strings.

```python
with open('file_example.txt', 'r') as file:
    lines = file.readlines()

print(lines)
```

Notice that the line separators are not deleted upon the lien splitting. As such, we need to strip the string first of the leading spaces.

```python
with open('file_example.txt', 'r') as file:
    lines = file.readlines()

clean_lines = []
for line in lines:
    clean_lines.append(line.strip())
print(clean_lines)
```

This can be rewritten using list comprehension.

```python
with open('file_example.txt', 'r') as file:
    lines = file.readlines()

clean_lines = [line.strip() for line in lines]
print(clean_lines)
```

#### Reading line by line

When you want to do the same thing to every line from the file cursor to the end of a file, you can iterate the file using a `for` loop. On each iteration, the file cursor is moved to the beginning of the next line.

```python
with open('file_example.txt', 'r') as file:
    for line in file:
        print(line)
```

```python
with open('file_example.txt', 'r') as file:
    lines = [line.strip() for line in file]

print(lines)
```

#### Using different methods for different contexts

Sometimes we want to treat lines differently depending on context. For example, if the files that you're working on has a header section followed by a series of records, you can use `readline()` to read one line at a time.

In this example, we'll be using the number of colored fox fur pelts produced in Hopedale, Labrador in the years 1834-1842.

```python
with open('hopedale.txt', 'r') as file:
    # Skip the description line
    data = file.readlines()
data
```

We can use `readline()` to skip the header and the different comment lines using a `while` loop. Then later on counting all the number of fox pelts in the duration of the year.

```python
with open('hopedale.txt', 'r') as file:
    # Skip the description line
    file.readline()

    # Keep reading and skipping comment lines
    data = file.readline().strip()
    while data.startswith('#'):
        data = file.readline().strip()

    # This is the first piece of data
    total_pelts = int(data)

    # Read the rest of the data
    for data in file:
        total_pelts += int(data)
```

Instead of aggregating the data, we might just want to get all the values per year of the data and possibly store it in a list or store it in a text file for later use.

```python
with open('hopedale.txt', 'r') as file:
    # Skip the description line
    file.readline()

    # Keep reading and skipping comment lines
    data = file.readline().strip()
    while data.startswith('#'):
        data = file.readline().strip()

    # This is the first piece of data
    pelts_data = [int(data)]

    # Read the rest of the data
    for data in file:
        pelts_data += [int(data)]
```

### Persisting Python Objects

We now have learned one way to store data, that is thru text files. However, sometimes what we want is to directly store a data object such as a dictionary, function, or even machine learning models for future use then recreate them when we want to. This process is called *serialization* - which means converting a data object into a format that allows storage or transmission and recreation through a reverse process called *deserialization*.

The `pickle` module implements Python's procols for serializing and de-serializing a Python object. "Pickling" is process of serialization while "unpickling" is the deserialization.

Most of the Python objects we know can be pickled. (see [*What can be pickled and unpickled?*](https://docs.python.org/3/library/pickle.html#what-can-be-pickled-and-unpickled) for more info).

These includes:

- constants - `None`, `True`, `False`, `Ellipsis`, and `NotImplemented`;
- integers, floating point numbers, complex numbers;
- strings, bytes, bytearrays
- tuples, lists, sets, and dictionaries;
- functions (except for `lambda` functions)
- classes accessible from the top level of a module

To serialize a python object we use the `pickle.dump()` function on an opened file to be written as bytes. For example, if we want to serialize the list `pelts_data` from awhile ago and store it to a pickle file `data/pelts_data.pickle`, we do:

### Multiple except clause

A `try` statement may have more than one except clause to specify the different handlers for different exceptions.

```python
try:
    raise Exception
except ValueError:
    print("There has been a value error")
except ZeroDivisionError:
    print("There has been a zero divison error")
```

### Accessing the argument of the exception

The *except* clause may specify a variable after the exception name. This variable is bound to the exception instance which typically has an `args` attribute that stores the arguments. You can access this arguments and use their value in your code for handling the exception.

```python
try:
    raise Exception("There is an error", "It is an exception")
except Exception as err:
    print(err)
    print(type(err))
    print(err.args)
    x, y = err.args
    print(x)
    print(y)
```

### The `else` block

We can also add an `else` block after the `try-except` statement. In this case, when present, it must follow all the `except` clauses. The `else` block works like this: Python attempts to run the code in the `try` block. If the `try` block was successful, the code goes in the `else` block. The `except` block tells Python what to do in case a certain exception arises when it tries to run the code in the `try` block.

For example we have the function `number_of_lines()` which takes in the a `filepath` then returns the number of lines in that file if successfully read. If not, it will print the message `"Cannot open the file [filepath]."`.

```python
def number_of_lines(filepath):
    try:
        file = open(filepath, 'r')
    except OSError:
        print(f"Cannot open the file {filepath}.")
    else:
        return len(file.readlines())
```

```python
number_of_lines('data/file_example.txt')
```

### Defining clean-up actions

If we want to perform actions that will be excecuted regardless as to whether an error has been raised or not. We can use the `finally` clause of the `try` statement.

When the `finally` clause is present, the finally clause will execute as the last task before the `try` statement completes.

The `try` statement has another optional clause which is intended to define clean-up actions that will be executed under all circumstances. 

```python
try:
    raise Exception
except ValueError:
    print("There has been a value error")
except ZeroDivisionError:
    print("There has been a zero divison error")
finally:
    print("Goodbye, World!")
```

Note the order in which the finally statement is evaluated - if a `try` statement reaches  a `break`, `continue`, or `return` statement, the `finally` clause will execute just prior to the break, continue, or return statement.

```python
def bool_return():
    try:
        return True
    finally:
        return False
```
