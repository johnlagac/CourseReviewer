# Sample Midterm (in-session)

## Context

When signing up for online sites, you are often asked to create a password that satisfies rules.  
These include, but are not limited to:
- Has a minimum number of characters.   
- Must have a mix of upper and lower case letters.  
- Must include at least one number.  
- Must include at least special character (e.g. "!" or "$").    

The reason for this is security: the more complex your password, the harder it is for malicious agents to guess your password through brute force or social engineering. 

The *[Password Game](https://neal.fun/password-game/)* is this idea taken in the direction of absurdity.  
A player supplies a password, and the game checks if it satisfies the current set of rules.  
Assuming they pass the current set, the game then adds a new rule and the user must update the password to pass the expanded set of rules.  
The difficulty of the game is both in the difficulty of some of the rules, as well as the complexity of multiple overlapping rules.

```python
# Example solution
def rule_0(pw: str) -> bool:
    """
    Rule to implement for problem 0 (example problem).

    Includes the following conditions:
        - Password must be at least 5 characters long
        - Password must contain at least 1 number
        - Password must contain at least 1 uppercase letter
        - Password must include a special character.

    Special characters include the following:
        ! @ # $ % ^ & * ( ) _ + - = [ ] { } | ; , . / < > ?

    Parameters
    ----------
    pw : str
        Password to check for conformity

    Returns
    -------
    out : bool
        Does the password pass all the conditions of this rule

    See Also
    --------
    <other file> : Some rules may need you to reference external files.

    """
    # check length
    out = (len(pw) >= 5)

    # check has uppercase
    out &= any(x.isupper() for x in pw)

    # check has number
    out &= any(x.isnumeric() for x in pw)

    # check special characters (list is specific to this question)
    doc_list = '! @ # $ % ^ & * ( ) _ + - = [ ] { } | ; , . / < > ?'
    spec_c = list(doc_list.replace(' ', ''))
    out &= any(x in spec_c for x in pw)

    return out
```
