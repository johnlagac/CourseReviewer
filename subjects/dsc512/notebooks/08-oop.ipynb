Python is an object-oriented programming language, which means that it provides features that support **object-oriented proramming**.

Up to now, most of the programs we have been writing use a procedural programming paradigm, in which, the focus is to write functions or procedures which operate on data. In OOP, the focus is on the creation of **objects** which contain both data and functionality together.

Usually each object is defined according to some object or concept in the real world, and the functions that operate on that object correspond to the ways real-world objects interact.

### Instances as return values

Functions and methods can return instances. For example, given two `Point` objects, find their midpoint.

````python
def midpoint(p1, p2):
    """Return the midpoint of points p1 and p2"""
    mx = (p1.x + p2.x) / 2
    my = (p1.y + p2.y) / 2

    return Point(mx, my)
````

````python
p = Point(3, 4)
q = Point(5, 12)
r = midpoint(p, q)
````

````python
r
````

````python
print(r)
````

We can do this as a method instead. For example, we wish to find the midpoint halfway between it and some other target point.

````python
class Point:
    """Point class represents and manipulates x, y coordinates"""

    def __init__(self, x=0, y=0):
        """Initialize a new point at x,y"""
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def halfway(self, target):
        """Return the halway point between self and the target"""
        mx = (self.x + target.x) / 2
        my = (self.y + target.y) / 2

        return Point(mx, my)

    def distance_from_origin(self):
        """Compute the distance of point to origin"""
        return ((self.x ** 2) + (self.y ** 2))**0.5
````

````python
p = Point(3, 4)
q = Point(5, 12)
r = p.halfway(q)
r
````

## 2 Polymorphism

Some languages, including Python, make it possible to have different meanings for the same operator when applied to different types. For example, `+` in Python means quite different things for integers and for strings. This feature is called **operator overloading**.

In our `Point` object, we can define how `+` operates by including an `__add__` method.

````python
4+4
````

````python
"4"+"4"
````

````python
class Point:
    """Point class represents and manipulates x, y coordinates"""

    def __init__(self, x=0, y=0):
        """Initialize a new point at x,y"""
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def halfway(self, target):
        """Return the halway point between self and the target"""
        mx = (self.x + target.x) / 2
        my = (self.y + target.y) / 2

        return Point(mx, my)

    def distance_from_origin(self):
        """Compute the distance of point to origin"""
        return ((self.x ** 2) + (self.y ** 2))**0.5

    def set_position(self, x, y):
        """Set the position of the point to a new x and y"""
        self.x = x
        self.y = y

    def move(self, dx, dy):
        """Move the object by the deltas"""
        self.x += dx
        self.y += dy
````

````python
p = Point(3, 4)
q = Point(5, 12)
r = p + q
````

````python
print(r)
````

We can do the same for multiplication.

````python
class Point:
    """Point class represents and manipulates x, y coordinates"""

    def __init__(self, x=0, y=0):
        """Initialize a new point at x,y"""
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __mul__(self, other):
        return self.x * other.x + self.y * other.y

    def __rmul__(self, other):
        return Point(other * self.x, other * self.y)

    def halfway(self, target):
        """Return the halway point between self and the target"""
        mx = (self.x + target.x) / 2
        my = (self.y + target.y) / 2

        return Point(mx, my)

    def distance_from_origin(self):
        """Compute the distance of point to origin"""
        return ((self.x ** 2) + (self.y ** 2))**0.5

    def set_position(self, x, y):
        """Set the position of the point to a new x and y"""
        self.x = x
        self.y = y

    def move(self, dx, dy):
        """Move the object by the deltas"""
        self.x += dx
        self.y += dy
````

````python
p = Point(3, 4)
q = Point(5, 12)
p * q
````

````python
r = 3 * p
print(r)
````

````python
p * 3
````

Most methods only work for a specific type, but there are certain operations that you want to apply to many types, such as arithmetic operations. If many types support the same set of operations, we can write functions that work on any of those types.

For example, consider `multadd` operation.

````python
def multadd(x, y, z):
    return x*y + z
````

This function will work for any values of x and y that can be multipleied and for any value of `z` that can be added to the product

````python
multadd(3, 2, 1)
````

````python
multadd(1, p, q)
````

````python
multadd(p, q, 2)
````

A function like this that can take arguments with different types is called **polymorphic**.

To determine whether a function can be applied to a new type, we apply Python's fundamental rule of polymorphism, called the **duck typeing rule**: *If all operations inside the function can be applied to the type, the function can be applied to the type.*

## 3 Inheritance

Giving one class the same methods as another is one way to make them polymorphic, but it suffers from repeated code if we always do that. If a programmer forgets just one line of code, the whole program can fail for reasons that will be difficult to track down.

A better approach is to use a feature called *inheritance*, which allows you to recycle code in yet another way.

Whenever you create a class using inheritance, you new class automatically inherits all of the attributs of the class *object*, much like a child inherits attributes from their parents.

For example, let's say we are managing people at a university. There are students and faculty. Both students and faculty have names, postal addresss, and email address; each student also has a student number, a list of course taken, and a list of courses he or she is currently taking. Each faculty member has a faculty number and a list of courses he or she is currently teaching.

We can define a `Faculty` and a `Student` class such that both of them are subclasses of a `Member` class which signifies that they are both members of a university.

````python
class Member:
    """A member of a university"""

    def __init__(self, name, address, email):
        self.name = name
        self.address = address
        self.email = email


class Faculty(Member):
    """A faculty member at a university"""

    def __init__(self, name, address, email, faculty_num):
        super().__init__(name, address, email)
        self.faculty_number = faculty_num
        self.courses_teaching = []


class Student(Member):
    """A student member at a university"""

    def __init__(self, name, address, email, student_num):
        super().__init__(name, address, email)
        self.student_number = student_num
        self.courses_taken = []
        self.courses_taking = []
````

````python
faculty_leo = Faculty('Leodegario Lorenzo II', 'Quezon City',
                      'llorenzo@aim.edu', 221039)
student_leo = Student('Leodegario Lorenzo II', 'Quezon City',
                      'llorenzo@aim.edu', 708472)
````
