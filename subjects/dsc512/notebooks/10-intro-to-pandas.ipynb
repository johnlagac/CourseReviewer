**Pandas** contains data structures and data manipulation tools designed to make data cleaning and analysis fast and convenient in Python. pandas is often used in tandem with numerical computing tools like NumPy and Scipy, analytical libraries like statsmodels and scikit-learn, and data visualization libraries like matplotlib.

pandas adopts significant parts of NumPy's idiomatic style of array-based computing, especially array-based functions and a preference for data processing without `for` loops. Although, the biggest difference between pandas and numpy is that pandas is designed for working with tabular or heterogeneous data. Whereas, NumPy is best at working with homogeneously typed numerical array data.


The common alias for numpy and pandas are `np` and `pd` respectively, which we will also use in our notebooks.

````python
import numpy as np
import pandas as pd
````

## 1 Introduction to pandas Data Structures

pandas has three fundamental data strucures: the `Series`, `DataFrame`, and `Index`. We will cover how each of these objects are created and manipulated in this section.

### `Series`

A series is a one-dimensional array-like object containing a sequence of values of the same type and an associated array of data labels called its *index*. The simplest `Series` is formed using an array of data:

````python
obj = pd.Series([4, 7, -5, 3])
obj
````

Since we did not specify an index, a default one consisting of integers `0` to `N - 1` (where `N` is the length of the data) is created.

````python
obj.array
````

````python
obj.index
````

Like with a NumPy array, data can be accessed by the associated integer indexed via the familiar Python square-bracket notation:

````python
obj[1]
````

````python
obj[:2]
````

#### Series as a Generalized NumPy Array

Without any explicit index, the `Series` look much like a one-dimensional array. We can add an explicit index definition to give the `Series` object additional capabilities. The index need not be an integer but can consist of any desired type. For example, we can use strings as an index:

````python
data = pd.Series([0.25, 0.5, 0.75, 1.0], index=['a', 'b', 'c', 'd'])
data
````

We can access the item as expected:

````python
data['b']
````

#### Series as Specialized Dictionary

We can think of a pandas Series a bit like a specialization of a Python dictionary. Whereas a Python dictionary maps arbitrary keys to a set of arbitrary values, a `Series` is a structure that maps typed keys to a set of typed values.

We can make this connection clear by defining a `Series` object from a Python dictionary.

````python
population_dict = {
    'California': 39_538_223,
    'Texas': 29_145_505,
    'Florida': 21_538_187,
    'New York': 20_201_249,
    'Pennsylvania': 13_002_700
}
population = pd.Series(population_dict)
population
````

Here, we can access item like a dictionary

````python
population['California']
````

At the same time, we can do array-style operations such as slicing:

````python
population['California': 'Florida']
````

#### Constructing Series Objects

We've seen a few ways of constructing a pandas `Series` from scratch. All of them are some version of the following:

```
pd.Series(data, index=index)
```

For example, data can be a list or NumPy array, in which case the default index  is an integer sequence:

````python
pd.Series([2, 4, 6])
````

Or data can be a scalar, in which it is repeated to fill the specified index:

````python
pd.Series(5, index=[100, 200, 300])
````

Or it can be a dictionary, in which case, `index` defaults to the dictionary keys.

````python
pd.Series({2: 'a', 1: 'b', 3: 'c'})
````

We can explicitly pass an index to control the order or the subset of keys used:

````python
pd.Series({2: 'a', 1: 'b', 3: 'c'}, index=[1, 2, 3])
````

Finally, note that we can give a series a name which will be handy if we would then want to transform the series into an index or a column in a data frame.

````python
my_series = pd.Series({2: 'a', 1: 'b', 3: 'c'}, index=[1, 2, 3])
my_series.name = 'sample'
my_series
````

````python
my_series = pd.Series({2: 'a', 1: 'b', 3: 'c'}, index=[1, 2, 3], name='sample')
my_series
````

### `DataFrame`

If a `Series` is an analog of one-dimensional array with explicit indices, a `DataFrame` is an analog of two-dimensional array with explicit row and column indices.

Just as you may think of a two-dimensional array as an ordered sequence of aligned one-dimensional columns, you can think of a `DataFrame` as a sequence of aligned `Series` objects. By aligned, we mean they share the same index.

To demonstrate this, we construct a new `Series` containing the area of each five states in our previous example.

````python
areas = [423_967, 695_662, 170_312, 141_297, 119_280]
states = ['California', 'Texas', 'Florida', 'New York', 'Pennsylvania']
````

````python
area_dict = {
    'California': 423_967,
    'Texas': 695_662,
    'Florida': 170_312,
    'New York': 141_297,
    'Pennsylvania': 119_280
}
area = pd.Series(area_dict)
area
````

We can use a dictionary to construct a two-dimensional object containing the information of both the population and area.

````python
states = pd.DataFrame({'population': population,
                       'area': area})
states
````

Like the `Series` object, the `DataFrame` has an `index` attribute that gives access to the index labels. 

````python
states.index
````

Additionally, it also has a `columns` attribute, which is an `Index` object holding the column labels.

````python
states.columns
````

Thus, `DataFrame` can be thought of a generalized two-dimensional array with both rows and columns having a genearlized index for accessing data.

#### DataFrame as Specialized Dictionary

We can also think of `DataFrame` as a specialization of a dictionary. Where a dictionary maps a key to a value, a `DataFrame` maps a column name to a `Series` of column data.

````python
states['area']
````

Notice a potential confusion here: in a two-dimensional NumPy array, `data[0]` will return the first *row*, whereas, `data['col0']` for a `DataFrame` will return the first column.

#### Constructing DataFrame Objects

A pandas `DataFrame` can be created in a variety of ways.

##### From a single `Series` object

````python
population.to_frame(name='population')
````

##### From a list of dicts

##### From a dictionary of Series objects

As we've seen before, we can create a DataFrame from a dictionary of `Series` objects:

##### From a two-dimensional NumPy array

Given a two-dimensonal array of data, we can create  a `DataFrame` with any specified column and index names. If omitted, an integer index will be used for each:

````python
pd.DataFrame(np.random.rand(3, 2), columns=['foo', 'bar'],
             index=['a', 'b', 'c'])
````

### `Index`

pandas' `Index` objects are responsible for holding the axis labels (including a `DataFrame`'s column names) and other metadata (like the axis names or names).

Any array you use when constructing a Series or DataFrame is internally converted to an `Index`:

````python
obj = pd.Series(np.arange(3), index=['a', 'b', 'c'])
obj
````

````python
index = obj.index
index
````

Index objects are immutable and thus can't be modified.

````python
index[0]
````

````python
index[0] = 'd'
````

This property makes it safer to share `Index` among data structures.

````python
obj2 = pd.Series(np.linspace(1, 3, 3), index=index)
obj2
````

````python
obj2.index is index
````

In addition to being array-like, an `Index` also behaves like a fixed-size set:

````python
states
````

````python
states.columns
````

````python
'population' in states.columns
````

````python
'Hawaii' in states.index
````

However, unlike sets, a pandas `Index` can contain duplicate labels:

````python
pd.Index(['foo', 'foo', 'bar', 'bar'])
````

Selections with duplicate labels will select all occurrences of that label.

## 2 Indexing and Selection

Accessing the data of `Series` and `DataFrame` will feel very familiar with how we index and select NumPy arrays. Although, there are few quirks to be aware of.

We'll start our discussion by considering the simplest case of the one-dimensional `Series` object, then move on to the more complicated two-dimensional `DataFrame` object.

### Data Selection in Series

#### Series as Dictionary

Like a dictionary, the `Series` object provides mapping from a collection of keys to a collection of values:

````python
data = pd.Series(np.linspace(0.25, 1, 4), index=['a', 'b', 'c', 'd'])
data
````

````python
data['b']
````

````python
'a' in data
````

````python
data.keys()
````

````python
list(data.items())
````

`Series` objects can also be modified with a dictionary-like syntax.

````python
data['e'] = 1.25
data
````

#### Series as One-Dimensional Array

A `Series` builds on this dictionary-like interface and provides array-style item selection via the same basic mechanisms as NumPy arrays - that is slices, masking, and fancy indexing.

````python
data['a':'c'] # slicing by explicit index
````

````python
data[0:2] # slicing by implicit index
````

````python
data[(data > 0.3) & (data < 0.8)] # boolean indexing
````

````python
data[['a', 'e']] # fancy indexing
````

Of these, slicing may be the most confusing, as with an explicit index, (e.g., `data['a':'c']`), the final index is *included* in the slice, while slicing with an implicit index (e.g., `data[0:2]`), the final index is *excluded* from the slice.

#### Indexers: `loc` and `iloc`

If your series has an explicit integer index, an indexing operation such as `data[1]` will use the explicit index, while slicing operation like `data[1:3]` will use the implicit index.

````python
data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
data
````

````python
data[1]
````

````python
data[1:3]
````

If you want to use explicit (or implicit) indexes, it is better to use the *indexer* attributes that pandas provides to avoid confusion.

First, the `loc` attribute allows indexing and slicing that always refers to the explicit index:

````python
data.loc[1]
````

````python
data.loc[1:3]
````

````python
data.iloc[1]
````

````python
data.iloc[1:3]
````

One guiding principle of Python code is that "explicit is better than implicit". The explicit nature of `loc` and `iloc` makes them helpful in maintaining clean and readable code.

### Data Selection in DataFrames

Recall that we showed that data frames act like a two-dimensional array and in other ways like a dictionary of `Series` structures sharing the same index. These analogies can be helpful to keep in mind as we explore data selection within this structure.

#### DataFrame as Dictionary

Let's consider once again the `states` DataFrame that we defined from the previous seciton.

````python
states
````

The individual `Series` that make up the DataFrame can be accessed via dictionary-style indexing of the column name:

````python
states['area']
````

Equivalently, we can use attribute-style access, with column names that are strings:

````python
states.area
````

Though this is a useful shorthand, this does not work for all cases.

For example, if the colujmn names are not strings, or if the column names conflict with methods of the DataFrame, this attribute-style access is not possible.

In particular, you should avoid the temptation to try column assignment via attributes. (i.e., use `data['pop'] = z` rather than `data.pop = z`

This dictionary-style syntax can also be used to modify the object. For example, creating a new column:

````python
states['density'] = states['population'] / states['area']
states
````

### DataFrame as Two-Dimensional Array

As mentioned previously, we can also view the `DataFrame` as a enhanced two-dimensional array. We can examine the raw underlying data array using the `to_numpy()` method

````python
states.to_numpy() # Creates a view of the data
````

````python
states.to_numpy(copy=True) # Creates a copy of the data
````

Methods such as transposition will also work in a data frame:

````python
states.T
````

For array style indexing, we can use again the `loc` and `iloc` indexers mentioned earlier. Using the `iloc` indexer, we can index the underlying array as if it were a simple NumPy array, but the DataFrame index and column labels are maintained in the result.

````python
states.iloc[:3, :2]
````

Similarly, using the `loc` indexer, we can index the underlying data in an array-like style but using the explicit index and column names:

````python
states.loc[:'Florida', :'population']
````

Any of the NumPy-style data access patterns can also be used within these indexers. For example, in the `loc` indexer, we can combine masking and fancy indexing as follows:

````python
states.loc[states.density > 120, ['population', 'density']]
````

Any of these indexing conventions may also be used to set or modify values; this is done in the standard way that you might be accustomed to from working with NumPy:

````python
states.iloc[0, 2] = 90
states
````

#### Additional Indexing Conventions

There are couple of extra indexing conventions that might seem at odds with the preceding discussion, but nevertheless can be useful in practice. First, while *indexing* refers to columns, *slicing* refers to rows:

````python
states['Florida':'New York']
````

Such slices can also refer to rows by number rather than by index

````python
states[1:3]
````

Similarly, direct masking operations are interpreted row-wise rather than column-wise

````python
states[states.density > 120]
````

#### Pitfalls with chained indexing

A common pitfall for new pandas users is to chain selections when assigning. For example, we want to clip the values of a specific column, we may imagine that we can do something like:

````python
states[states.density > 100]['density'] = 100
````

However, instead we see a warning, which tells us that we are trying to modify a temporary value (the nonempty result of `states[states.density > 100]`) instead of the original dataframe, which we might have been intending. Hence, notice that the original data frame was not modified.

````python
states
````

The fix, is to rewrite the chained assignment to use a single `loc` operation:

````python
states.loc[states.density > 100, 'density'] = 100
states
````

### Reindexing

Reindexing means creating a new object with the values rearranged to align with the new desired index. For example, given a series `obj`,

````python
obj = pd.Series([4.5, -7.2, -5.3, 3.6], index=['d', 'b', 'a', 'c'])
obj
````

Calling the `reindex` arranges the data according to the new index, adding null values if they were not present originally:

````python
obj.reindex(['a', 'b', 'c', 'd', 'e'])
````

If you want to fill the null values, you may do so by specifying a method for interpolation.

````python
obj2 = pd.Series(['blue', 'purple', 'yellow'], index=range(0, 5, 2))
obj2
````

````python
obj2.reindex(range(6))
````

````python
obj2.reindex(range(6), method='ffill')
````

````python
obj2.reindex(range(6), method='bfill')
````

You can choose to alter the row, columns, or both by specifying the appropriate arguments.

````python
states
````

````python
states.reindex(index=['Texas', 'New Jersey', 'California'],
               columns=['population', 'density', 'happiness'])
````

### Arithmetic and Data Alignment

pandas can make it much simpler to work with objects that have different indexes.

For example, let's say we're given a data containing accumulated purchases of users:

````python
np.random.seed(1337)
data = pd.DataFrame({'user_id': ['001', '002', '023'],
                     'milk': [0, 1, 2],
                     'cheese': [0, 0, 1],
                     'diaper': [1, 0, 1],
                     'beer': [1, 1, 0]}).set_index('user_id')
data.columns.name = 'product'
data
````

Then, we wish to update the data with a newly acquired data:

````python
incoming_data = pd.DataFrame({'user_id': ['001', '002', '007'],
                              'beer': [1, 0, 1],
                              'chips': [1, 1, 1],
                              'pew-pew': [0, 0, 1]}).set_index('user_id')
incoming_data.columns.name = 'product'
incoming_data
````

Adding these returns a DataFrame with index and columns that are unions of the ones in each data frame

````python
data + incoming_data
````

If we want to ensure alignment between the two data frames, we can use the corresponding arithmetic method with the specified fill value.

````python
data.add(incoming_data, fill_value=0)
````

Notice that we still have `NaN`s for locations with missing values for both dataframe. To also fill this, we could use instead the `fillna` value specifically of the resulting data frame.

````python
updated_data = data.add(incoming_data, fill_value=0).fillna(0)
updated_data
````

#### Operations between `DataFrame` and `Series`

We can also broadcast using pandas. By default, arithmetic between DataFrame and Series matches the index of the Series on the columns of the DataFrame, broadcasting down the rows:

````python
one_bucket_for_everybody = pd.Series({'beer': 6})
one_bucket_for_everybody
````

````python
updated_data + one_bucket_for_everybody
````

````python
updated_data.add(one_bucket_for_everybody, fill_value=0)
````

````python
updated_data.add(
    pd.DataFrame(6, index=updated_data.index, columns=['beer']),
    fill_value=0)
````

````python
updated_data
````

````python
updated_data['beer'] += 6
updated_data
````

### Function Application and Mapping

Numpy ufuncs also work with pandas objects:

````python
np.sign(updated_data)
````

````python
np.square(updated_data)
````

Another frequent operation is applying a function on one-dimensional arrays to each column or row. For example, we want to get the different between the minimum and maximum values in our data frame:

````python
def maxmin_diff(x):
    """Return the difference between the max and min of an array"""
    return x.max() - x.min()
````

````python
updated_data
````

Across rows

````python
updated_data.apply(maxmin_diff)
````

Across columns

````python
updated_data.apply(maxmin_diff, axis='columns')
````

The function need not return a scalar value, it can also return a `Series` with multiple values.

````python
def min_max_mean(x):
    """Return the min, max, and mean of an array"""
    return pd.Series([x.min(), x.max(), x.mean()],
                     index=['min', 'max', 'mean'])
````

````python
updated_data.apply(min_max_mean)
````

````python
updated_data.apply(min_max_mean, axis=1)
````

You can also apply any function element-wise using the `applymap` method.

For example, we want to convert the data type of the values of our data frame to `int`.

````python
updated_data
````

We cannot do it by directly using the `int` function.

````python
int(updated_data)
````

Instead, we use `applymap` to apply the `int` function to each element in the data frame.

````python
updated_data = updated_data.applymap(int)
updated_data
````

### Sorting and Ranking

Sorting the dataset by some criterion is another important built-in operation.

````python
updated_data.sort_index(ascending=False)
````

````python
updated_data.sort_index(axis='columns', ascending=False)
````

If we want to sort according to the values of a Series or a column in a DataFrame, we use `sort_values`.

````python
updated_data.milk.sort_values()
````

````python
updated_data.sort_values(by='milk')
````

*Ranking* assigns rank from one through the number of valid data points in an array starting from the lowest value. This can be applied for both `Series` and `DataFrame`.

For example, we want to rank the top consumer for each of the product, we can do:

````python
updated_data.rank(ascending=False)
````

If we want to handle ties in ranking, we specify the `method` argument.

````python
updated_data.rank(ascending=False, method='first')
````

We can also rank across columns. Here, we get which product does the user consumes the most.

````python
updated_data.rank(axis='columns', ascending=False, method='min')
````
